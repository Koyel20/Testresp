```python
#%pip install -q scikit-learn

paragraph = '''ham  Go until jurong point, crazy.. Available only 
ham                      Ok lar... Joking wif u oni
spam  Free entry in 2 a wkly comp to win FA Cup fina
ham  u dun say so early hor... u c already then say                     
spam  This is the 2nd time we have tried 2 contact u
ham               Will u b going to esplanade fr home?
ham  Pity, * was in mood for that. So...any other
ham  The guy did some bitching but I tacted like i'd
ham  Nah I don't think he goes to usf, he lives around 
ham                         Rofl. Its true to its name'''

import re 
import nltk
import pandas as pd
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer
# Build the messages DataFrame from the in-memory `paragraph` string
corpus = []
lines = [ln for ln in paragraph.splitlines()]
data = []
for ln in lines:
	parts = ln.strip().split(None,1)  # split into label and the rest of the message
	if len(parts) == 2:
		label, msg = parts
	else:
	
		label =  parts[0]
		msg = ''
	data.append((label, msg))
messages = pd.DataFrame(data, columns=['label', 'message'])
#nltk.download('stopwords')
corpus = []
for i in range(0, len(messages)):
    # Remove all the special characters
    review = re.sub('[^a-zA-Z]', ' ', messages['message'][i])
    # Convert to lowercase
    review = review.lower()
    # Tokenize
    review = review.split()
    # Remove stopwords
    review = [nltk.stem.WordNetLemmatizer().lemmatize(word) for word in review if word not in stopwords.words('english')]
##create bag in word model
    review = ' '.join(review)
	
    corpus.append(review)
print(corpus)
cv = CountVectorizer()
print(cv.fit_transform(corpus))

# Get the feature names (vocabulary in order)
print(cv.get_feature_names_out())
cv = CountVectorizer(binary=True, max_features=20)

X = cv.fit_transform(corpus).toarray()
features = cv.get_feature_names_out()
counts = X.sum(axis=0)  # total occurrences per token across all documents
max_count = counts.max()
max_tokens = features[counts == max_count]
print(X)
print("Features:", list(features))
print("Counts:", dict(zip(features, counts)))
print("Max count:", int(max_count))
print("Tokens with max count:", list(max_tokens))
print(X.shape)



```
