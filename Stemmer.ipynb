```python
from nltk.stem  import *
words = ["eating","sleeping","eaten","Congratulation","gone","history"]
prot = PorterStemmer()


for root in words:
    print(f"{root}--->{prot.stem(root)}")

reg_stemmer = RegexpStemmer('ing$|e$|s$|able$',min = 4)

for root in words:
    print(f"Regexp stemmer :{root}--->{reg_stemmer.stem(root)}")

snowball_stemmer = SnowballStemmer( language= 'english' )
for root in words:
    print(f"Snowball stemmer :{root}--->{snowball_stemmer.stem(root)}")
'''porter'''
print(prot.stem('fairly')) 
print(prot.stem('sportingly'))
'''regexp'''
print(reg_stemmer.stem('fairly')) 
print(reg_stemmer.stem('sportingly'))
'''snowball'''
print(snowball_stemmer.stem('fairly')) 
print(snowball_stemmer.stem('sportingly'))
```
