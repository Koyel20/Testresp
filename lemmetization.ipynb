```python
import nltk
lemmetizer = nltk.stem.WordNetLemmatizer()
## words = ["eating","sleeping","eaten","Congratulations","fairly","gone","history"]
##for root in words:
  ##  print(f"{root}--->{lemmetizer.lemmatize(root, pos = 'a')}")
##synsets = nltk.corpus.wordnet.synsets('happy')
##print(synsets)

##nltk.download('stopwords')
##nltk.corpus.stopwords.words('english')
## paragraphs = ''' Rapid change has an increasingly large impact on the environment. Deserts are expanding, while heat waves and wildfires are becoming more common. Amplified warming in the Arctic has contributed to thawing permafrost, retreat of glaciers and sea ice decline. Higher temperatures are also causing more intense storms, droughts, and other weather extremes. Rapid environmental change in mountains, coral reefs, and the Arctic is forcing many species to relocate or become extinct. '''
paragraph = ''' this is koyel mandal from India. I love chinese food'''
sentence = nltk.sent_tokenize(paragraph)
for line in sentence:
    line_words = nltk.word_tokenize(line)

   
    each_word = [lemmetizer.lemmatize(word) for word in line_words if word not in set(nltk.corpus.stopwords.words('english'))]
    
    sentence[sentence.index(line)] = ' '.join(each_word)
pos_tags = nltk.pos_tag(sentence)
print("this is a test for split() function".split())
print(pos_tags)

##Pos_tags(parts of speech tagging)
#nltk.download('averaged_perceptron_tagger')
##Name entity recognition
#nltk.download('maxent_ne_chunker_tab')
#nltk.download('words')
##Name entity recognition
print(nltk.ne_chunk(pos_tags))
```
